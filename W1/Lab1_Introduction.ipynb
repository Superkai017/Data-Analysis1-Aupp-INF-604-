```python
data['Cabin'].isna().sum() * 100 / data.shape[0]
```

```python
data.drop(columns=['Cabin'], inplace=True)
```

```python
# Fill missing Age with the median (median, not mean: Age is skewed by a few old passengers)
median_age = data['Age'].median()
print(f"Missing before          : {data['Age'].isna().sum()}")

data['Age'] = data['Age'].fillna(median_age)   # assign the result, do NOT add inplace=True

print(f"Missing after           : {data['Age'].isna().sum()}")
data['Age'].describe()
```

```python

mode_port = data['Embarked'].mode()[0]
print(f"Missing before         : {data['Embarked'].isna().sum()}")

data['Embarked'] = data['Embarked'].fillna(mode_port)  

print(f"Missing after          : {data['Embarked'].isna().sum()}")
data['Embarked'].value_counts()
```

```python
#I How many 1st class passengers survived? How about 2nd and 3rd class passengers? 
data.groupby('Pclass')['Survived'].sum()
```

```python
#J.** How many female passengers survived? How many males did?
data.groupby('Sex')['Survived'].sum()
```

```python
#**K.** How many people from each embarkation port survived?
by_port = data.groupby('Embarked')['Survived'].sum()
pd.crosstab(data['Embarked'], data['Survived'])
```

```python
#L. Was `Jack` on the ship? How about `Rose`?
jack= data[data['Name'].str.contains('Jack', case=False)]
rose= data[data['Name'].str.contains('Rose', case=False)]


print(f"Names containing 'Jack' : {len(jack)}")
print(f"Names containing 'Rose' : {len(rose)}")
display(jack[['Name', 'Sex', 'Age', 'Pclass', 'Survived']])
display(rose[['Name', 'Sex', 'Age', 'Pclass', 'Survived']])
```

```python
# Those 3 hits are only substrings: Jackson, Ambrose, Rosen.
# \b marks a word boundary, so the name has to stand on its own instead of
# sitting inside a longer word. r'...' keeps the backslash literal for the regex.
jack_w = data[data['Name'].str.contains(r'\bJack\b', case=False, regex=True)]
rose_w = data[data['Name'].str.contains(r'\bRose\b', case=False, regex=True)]

print(f"Passengers actually named Jack : {len(jack_w)}")
print(f"Passengers actually named Rose : {len(rose_w)}")

# The movie characters' surnames are not in the data either
for surname in ['Dawson', 'DeWitt', 'Bukater']:
    print(f"{surname:8}: {data['Name'].str.contains(surname, case=False).sum()} match(es)")
```

`Answer:` **No — neither Jack nor Rose was on the ship.**

Searching the plain text `Jack` and `Rose` returns 3 passengers, but all three are substring matches, not people with those names:

- `Brewe, Dr. Arthur Jackson` — *Jack* inside **Jack**son
- `Hood, Mr. Ambrose Jr` — *rose* inside Amb**rose**
- `Aks, Mrs. Sam (Leah Rosen)` — *Rose* inside **Rose**n

Adding a word boundary (`\b`) forces the name to stand on its own, and that returns **0 for both**. The characters' surnames (Dawson, DeWitt Bukater) do not appear either. Jack and Rose were invented for the 1997 film — they were never real passengers.
